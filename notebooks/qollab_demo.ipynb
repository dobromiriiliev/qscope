# Q-Scope
Adaptive Hamiltonian drift diagnosis with Qiskit.

```python
from pathlib import Path
import sys

root = Path.cwd()
if not (root / "src" / "qscope").exists() and (root.parent / "src" / "qscope").exists():
    root = root.parent
source = root / "src"
if str(source) not in sys.path:
    sys.path.insert(0, str(source))

from qscope.config import load_config
from qscope.factory import build_components, build_engine
from qscope.benchmark import run_benchmark, summarize
```

```python
cfg = load_config("configs/mvp.json")
nominal, hypotheses, experiments, predictions, likelihoods = build_components(cfg)
print(len(hypotheses), "hypotheses")
print(len(experiments), "experiments")
```

```python
result = build_engine(cfg, nominal, hypotheses, experiments, predictions, likelihoods).run()
result.to_dict()
```

```python
rows = run_benchmark(cfg, nominal, hypotheses, experiments, predictions, likelihoods, trials=1, output="artifacts/qollab_benchmark.csv")
summarize(rows)
```
